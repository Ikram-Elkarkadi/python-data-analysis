```python

# Étape 1 : Installation et importation des bibliothèques

!pip -q install ucimlrepo

import pandas as pd
from ucimlrepo import fetch_ucirepo

# Étape 2 : Récupération du dataset officiel UCI

student_performance = fetch_ucirepo(id=320)

# Étape 3 : Récupération des données
features = student_performance.data.features
targets = student_performance.data.targets

# Étape 4 : Regroupement des variables dans un seul DataFrame
df = pd.concat([features, targets], axis=1)

# Étape 5 : Affichage des premières lignes
print("Dimensions du dataset :", df.shape)
print("\nAperçu des données :")
display(df.head())

# Étape 6 : Informations générales
print("\nInformations sur les colonnes :")
df.info()
```

```python

# Étape 1 : Dimensions du dataset
print("Nombre de lignes et de colonnes :", df.shape)

# Étape 2 : Afficher les noms des colonnes
print("\nNoms des colonnes :")
print(df.columns.tolist())

# Étape 3 : Afficher les types des variables
print("\nTypes des variables :")
print(df.dtypes)

# Étape 4 : Statistiques descriptives
print("\nStatistiques descriptives :")
display(df.describe(include="all").T)
```

```python

# Étape 1 : Vérifier les valeurs manquantes
missing_values = df.isnull().sum()

print("Valeurs manquantes par colonne :")
display(missing_values[missing_values > 0])

# Étape 2 : Calculer le pourcentage de valeurs manquantes
missing_percentage = (df.isnull().mean() * 100).round(2)

print("\nPourcentage de valeurs manquantes :")
display(missing_percentage[missing_percentage > 0])

# Étape 3 : Vérifier les lignes dupliquées
duplicates = df.duplicated().sum()

print("\nNombre de lignes dupliquées :", duplicates)
```

```python

# Créer une copie pour conserver les données originales
df_clean = df.copy()

# Supprimer les lignes entièrement dupliquées, si elles existent
df_clean = df_clean.drop_duplicates()

print("Dimensions initiales :", df.shape)
print("Dimensions après nettoyage :", df_clean.shape)

print("\nValeurs manquantes après nettoyage :")
print(df_clean.isnull().sum().sum())
```

```python

# Vérifier les colonnes disponibles dans le dataset
print("Colonnes disponibles :")
print(df_clean.columns.tolist())

# Vérifier les dimensions du dataset
print("\nDimensions du dataset :", df_clean.shape)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Définir le style des graphiques
sns.set_theme(style="whitegrid")

# Créer la figure
plt.figure(figsize=(9, 5))

# Tracer la distribution des notes finales
sns.histplot(data=df_clean, x="G3", bins=11, kde=True)

# Ajouter les titres et les labels
plt.title("Distribution des notes finales des étudiants",
          fontsize=14)
plt.xlabel("Note finale (G3 / 20)")
plt.ylabel("Nombre d'étudiants")

plt.tight_layout()
plt.show()
```

```python

# Calculer la moyenne de G3 pour chaque catégorie de studytime
mean_by_studytime = (
    df_clean.groupby("studytime")["G3"]
    .mean()
    .reset_index()
)

# Afficher le tableau des résultats
mean_by_studytime["G3"] = mean_by_studytime["G3"].round(2)
display(mean_by_studytime)

# Créer le graphique
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_clean,
    x="studytime",
    y="G3",
    hue="studytime",
    legend=False,
    errorbar=None
)

plt.title("Note finale moyenne selon le temps d'étude")
plt.xlabel("Catégorie de temps d'étude (1 à 4)")
plt.ylabel("Note finale moyenne (/20)")

plt.tight_layout()
plt.show()
```

```python

# Créer un nuage de points
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df_clean,
    x="absences",
    y="G3",
    alpha=0.65
)

plt.title("Relation entre les absences et les notes finales")
plt.xlabel("Nombre d'absences")
plt.ylabel("Note finale (G3 / 20)")

plt.tight_layout()
plt.show()
```
