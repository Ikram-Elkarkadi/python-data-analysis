```python

# Étape 1 : Installation et importation des bibliothèques

!pip -q install ucimlrepo

import pandas as pd
from ucimlrepo import fetch_ucirepo

# Étape 2 : Récupération du dataset officiel UCI

student_performance = fetch_ucirepo(id=320)

# Étape 3 : Récupération des données
features = student_performance.data.features
targets = student_performance.data.targets

# Étape 4 : Regroupement des variables dans un seul DataFrame
df = pd.concat([features, targets], axis=1)

# Étape 5 : Affichage des premières lignes
print("Dimensions du dataset :", df.shape)
print("\nAperçu des données :")
display(df.head())

# Étape 6 : Informations générales
print("\nInformations sur les colonnes :")
df.info()
```

```python

# Étape 1 : Dimensions du dataset
print("Nombre de lignes et de colonnes :", df.shape)

# Étape 2 : Afficher les noms des colonnes
print("\nNoms des colonnes :")
print(df.columns.tolist())

# Étape 3 : Afficher les types des variables
print("\nTypes des variables :")
print(df.dtypes)

# Étape 4 : Statistiques descriptives
print("\nStatistiques descriptives :")
display(df.describe(include="all").T)
```

```python

# Étape 1 : Vérifier les valeurs manquantes
missing_values = df.isnull().sum()

print("Valeurs manquantes par colonne :")
display(missing_values[missing_values > 0])

# Étape 2 : Calculer le pourcentage de valeurs manquantes
missing_percentage = (df.isnull().mean() * 100).round(2)

print("\nPourcentage de valeurs manquantes :")
display(missing_percentage[missing_percentage > 0])

# Étape 3 : Vérifier les lignes dupliquées
duplicates = df.duplicated().sum()

print("\nNombre de lignes dupliquées :", duplicates)
```

```python

# Créer une copie pour conserver les données originales
df_clean = df.copy()

# Supprimer les lignes entièrement dupliquées, si elles existent
df_clean = df_clean.drop_duplicates()

print("Dimensions initiales :", df.shape)
print("Dimensions après nettoyage :", df_clean.shape)

print("\nValeurs manquantes après nettoyage :")
print(df_clean.isnull().sum().sum())
```
